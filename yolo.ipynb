```python
from ultralytics import YOLO

# Load a pretrained YOLO11n model
yolo_model = YOLO("yolo11n.pt")
```

```python
results = yolo_model("/content/Screenshot 2025-08-12 144159.png")  # Predict on an image
results[0].show()  # Display results
```

```python
from ultralytics import YOLOE

# Initialize a YOLOE model
model = YOLOE("yoloe-v8s-seg.pt")  # or select yoloe-11s/m-seg.pt for different sizes

# Set text prompt
names = ["paino" , "person"]
model.set_classes(names, model.get_text_pe(names))

# Execute prediction for specified categories on an image
results = model.predict("/content/Screenshot 2025-08-22 003123.png")

# Show results
results[0].show()
```
